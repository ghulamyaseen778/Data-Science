```python
questions=[
    {
        "desc":"How was the food quality?",
        "rating":0,
        "id":"q1",
    },
    {
        "desc":"How was the customer service?",
        "rating":0,
        "id":"q2",
    },
    {
        "desc":"How clean was the restaurant?",
        "rating":0,
        "id":"q3",
    },
    {
        "desc":"How was the price of the food?",
        "rating":0,
        "id":"q4",
    },
    {
        "desc":"How was your overall experience?",
        "rating":0,
        "id":"q5",
    },

]
```

```python
def checkRating(rating):
  try:
    rate = int(rating)
    if type(int(rate)) != int:
      return {"error":True,"msg":"Enter correct number"}
    if rate<1 or rate>5:
      return {"error":True,"msg":"Enter select one option from given options"}

    return {"error":False,"msg":"Thank you for response","rating":rate}

  except:
       return {"error":True,"msg":"Internal error try again"}

```

```python
def ratingFomrate(ques):
  try:
    selectedRating=0
    print(ques)
    print("1 - Poor, 2 - Average, 3 - Good, 4 - Very Good, 5 - Excellent. ")
    selectedRating = input("Select rating from these options:")
    selectedRating = checkRating(selectedRating)
    # print(selectedRating,"selectedRating")
    while selectedRating["error"]:
      print(selectedRating["msg"])
      selectedRating = input("Select rating from these options:")
      selectedRating = checkRating(selectedRating)
    print(selectedRating["msg"])
    return {"error":False,"msg":"","rating":selectedRating["rating"]}
  except err:
    print(err)
    return {"error":True,"msg":"Internal error try again"}
```

```python
def totalRatingComments(rating):
  if rating>=4.0:
    print("Excellent Experience")
  elif rating >= 3.0 or rating<4.0:
     print("Good Experience")
  elif rating >= 2.0 or rating<3.0:
     print(" Average Experience")
  elif rating<2.0:
     print("Average Experience")
```

```python
totalRating = 0

for ind,ques in enumerate(questions):
  # print(ques,'ques')
  print("---------------")
  rate = ratingFomrate(ques["desc"])
  # print(rate,"rate")
  if rate["error"]:
    print("Internal error")
  else:
    questions[ind]["rating"] = rate["rating"]
    totalRating += rate["rating"]

print("---------------")
print(f"Total Rating: {totalRating}. Average Rating: {totalRating / int(len(questions))}")
totalRatingComments(totalRating / int(len(questions)))
```
